```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_sales_raw_messy.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors="coerce")
```

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors="coerce",dayfirst=False)
```

```python
print(df["Order_Date"].isnull().sum())
```

```python
print(df["City"].unique())
```

```python
df["City"]=df["City"].fillna("Unknown")
```

```python
city_mapping={
    "Banglore":"Bengaluru",
    "bengaluru":"Bengaluru",
    "BENGALURU":"Bengaluru",
    "hyderabad":"Hyderabad",
    "HYDERABAD":"Hyderabad",
    "Hyd":"Hyderabad",
    "mumbai":"Mumbai",
    "MUMBAI":"Mumbai",
    "chennai":"Chennai",
    "CHENNAI":"Chennai",
    "kolkata":"Kolkata",
    "CALCUTTA":"Kolkata",
    "new delhi":"New Delhi",
    "NEW DELHI":"New Delhi",
    "NewDelhi":"New Delhi"
}
df["City"]=df["City"].replace(city_mapping)
```

```python
print(df["Category"].unique())
```

```python
category_mapping={
    "electronics":"Electronics",
    "fashion":"Fashion",
    "beauty":"Beauty",
    "sports":"Sports",
    "Home and Kitchen":"Home & Kitchen"
}
df["Category"]=df["Category"].replace(category_mapping)
```

```python
print(df["Gender"].isnull().sum())
```

```python
df["Gender"]=df["Gender"].fillna("Unknown")
```

```python
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
```

```python
df["Discount"]=pd.to_numeric(df["Discount"],errors="coerce")
```

```python
df["Discount"]=df["Discount"].fillna(0)
```

```python
print(df[df["Quantity"]<=0])
```

```python
df=df[df["Quantity"]>0]
```

```python
print(df[df["Unit_Price"]<=0])
```

```python
df = df[df["Unit_Price"] > 0]
```

```python
df["Revenue"]=df["Revenue"].astype(str)
```

```python
df["Revenue"]=df["Revenue"].str.replace("₹", "", regex=False)
df["Revenue"]=df["Revenue"].str.replace(",", "", regex=False)
```

```python
df["Revenue"]=pd.to_numeric(df["Revenue"],errors="coerce")
```

```python
print(df["Revenue"].isnull().sum())
```

```python
df["Calculated_Revenue"]=(df["Quantity"]*df["Unit_Price"]*(1-df["Discount"]))
```

```python
df["Revenue_Difference"] = (
    df["Revenue"] -
    df["Calculated_Revenue"]
)
```

```python
df = df.drop(
    columns=[
        "Calculated_Revenue",
        "Revenue_Difference"
    ]
)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Rows involved in duplicates:", df[df.duplicated(keep=False)].shape[0])
```

```python
duplicate_orders = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

print(duplicate_orders[
    ["Order_ID", "Order_Date", "Customer_ID",
     "Product_ID", "Quantity", "Revenue"]
])
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicates after removal:", df.duplicated().sum())
```

```python
duplicate_order_ids = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

print(duplicate_order_ids[
    ["Order_ID", "Order_Date", "Customer_ID",
     "Product_ID", "Product_Name", "Quantity",
     "Unit_Price", "Discount", "Revenue", "Order_Status"]
])
```

```python
print(df.describe())
```

```python
df = df.copy()
```

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.strftime("%B")
df["Quarter"] = df["Order_Date"].dt.quarter
df["Day_Name"] = df["Order_Date"].dt.strftime("%A")
```

```python
def age_group(age):
    if age <= 25:
        return "18-25"
    elif age <= 35:
        return "26-35"
    elif age <= 45:
        return "36-45"
    elif age <= 55:
        return "46-55"
    else:
        return "56+"

df["Age_Group"] = df["Age"].apply(age_group)
```

```python
print(df["Order_Date"].dtype)
```

```python
print(df[["Order_Date", "Year", "Month", "Month_Name", "Quarter", "Day_Name"]].head())
```

```python
print("===== FINAL DATA QUALITY CHECK =====")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())

print("\nInvalid quantities:")
print((df["Quantity"] <= 0).sum())

print("\nInvalid prices:")
print((df["Unit_Price"] <= 0).sum())

print("\nInvalid revenue:")
print((df["Revenue"] <= 0).sum())
```
